```python
# prime number
def is_prime(n):
    for i in range(2, n):
        if(n%i==0):
            print("not a prime number")
            break
    else:
        print("prime number")
is_prime(5)
```

```python
#factorial
def factorial(n):
    for i in range(1,n):
        n=n*i
    print(n)
factorial(5)

```

```python
# amstrong number
def amstrong(n):
    sum=0
    temp=n
    while temp>0:
        digit=temp%10
        print(digit)
        sum+=digit**3
        temp//=10
        print(temp)
    return sum==n
amstrong(153)
```

```python
# decimal to binary using while loop/ for loop
def decimal_to_binary(n):
    binary=''
    while n>0:
        rem=n%2
        # print(rem)
        binary+=str(rem)
        n=n//2
        print(n)
    return binary[::-1]
decimal_to_binary(10)

```

```python
#fibonacci series
def fibonacci(n):
    a=0
    b=1
    if n==1:
        print(a)
    elif n==2:
        print(a,b)
    else:
        print(a,b,end=" ")
        for i in range(n):
            c=a+b
            a=b
            b=c
            print(b,end=" ")
fibonacci(10)
```

```python
#any trick question to find the number of digits in a number
def digits(n):
    count=0
    while n>0:
        n=n//10
        count+=1
    return count
digits(123456789)
```

```python
def digits(n):
    count=0
    for i in range(len(str(n))):
        count+=1
    return count
digits(123456789)
```

```python
#faltten a  nested list
def flatten_list(nested_list):
    flattened_list=[]
    for item in nested_list:
        if isinstance(item,list):
            flattened_list.extend(flatten_list(item))
        else:
            flattened_list.append(item)
    return flattened_list
nested_list=[1,2,3,[4,5,6],[7,8,9]]
print(flatten_list(nested_list))

```

```python
# flatten a nested list using itertools.chain and recursion
from collections.abc import Iterable

def flatten_list(nested_list):
    for item in nested_list:
        if isinstance(item, Iterable) and not isinstance(item, (str, bytes)):
            yield from flatten_list(item)
        else:
            yield item

nested_list = [1, 2, 3, [4, 5, 6], [7, 8, 9]]
print(list(flatten_list(nested_list)))
```

```python
# need to find the longest word in the string
def word(s):
    l=s.split()
    size= {}
    for i in l:
        size[i]=len(i)
    return max(size,key=size.get)
word("Python is very powerfull language very")
```
