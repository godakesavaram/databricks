```python
data = [
    (1, 300, "31-Jan-2021"),
    (1, 400, "28-Feb-2021"),
    (1, 200, "31-Mar-2021"),
    (2, 1000, "31-Oct-2021"),
    (2, 900, "31-Dec-2021")
]
df = spark.createDataFrame(data, ["empid", "commissionamt", "monthlastdate"])
df.show()
```

```python
df=df.withColumn("monthlastdate", to_date(col("monthlastdate"), "dd-MMM-yyyy"))
```

```python
from pyspark.sql.window import Window

win=Window.partitionBy("empid").orderBy(desc("monthlastdate"))
```

```python
fin=df.withColumn("rank", row_number().over(win)).filter(col("rank")==1)
fin.withColumn("monthlastdate", date_format(col("monthlastdate"), "dd-MMM-yyyy")).show()
```

```python
df = spark.createDataFrame([("Renuka1992@gmail.com", "9856765434"), ("anbu.arasu@gmail.com", "9844567788")], ["email", "mobile"])
df.show()
```

```python
from pyspark.sql.functions import col, expr

# Mask middle letters of email, keeping first and last character before '@'
df_masked = df.withColumn(
    "masked_email",
    expr("""
        concat(
            substring(email, 1, 1),
            repeat('*', length(split(email, '@')[0]) - 2),
            substring(email, length(split(email, '@')[0]), 1),
            '@',
            split(email, '@')[1]
        )
    """)
)
display(df_masked)
```

```python
from pyspark.sql.functions import explode, split

# Sample DataFrame with a column 'text'
data = [("Hello world",), ("Hello Databricks notebook",), ("Word count logic",)]
df = spark.createDataFrame(data, ["text"])
df.show(truncate=False)
df.printSchema()

# Split text into words and explode
words_df = df.select(explode(split(col("text"), " ")).alias("word"))
words_df.show(truncate=False)

# Group by word and count occurrences
word_count_df = words_df.groupBy("word").count()

display(word_count_df)
```
